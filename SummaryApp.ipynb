```python
import pandas as pd
from transformers import T5Tokenizer, Trainer, TrainingArguments, T5ForConditionalGeneration
```

```python
train_data=pd.read_csv("/samsum-train.csv", on_bad_lines='skip');
valid_data=pd.read_csv("/samsum-validation.csv", on_bad_lines='skip');
```

```python
train_data.head(5)
```

```python
train_data=train_data.sample(n=4000,random_state=42).reset_index(drop=True)
valid_data=valid_data.sample(n=500,random_state=42).reset_index(drop=True)
```

```python
#Data pre processing
```

```python
import re

def clean_data(text):
  text=re.sub(r"\r\n"," ",text) # lines
  text=re.sub(r"\s+"," ",text) #spaces
  text=re.sub(r"<.*?>"," ",text) # html tags
  text.strip().lower()

  return text
```

```python
train_data["dialogue"]=train_data["dialogue"].apply(clean_data)
train_data["summary"]=train_data["summary"].apply(clean_data)

valid_data["dialogue"]=valid_data["dialogue"].apply(clean_data)
valid_data["summary"]=valid_data["summary"].apply(clean_data)
```
